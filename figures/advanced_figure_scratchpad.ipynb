```python
import numpy as np
import matplotlib.pyplot as plt


num_qubits = np.arange(7, 15) # 8 data points

plt.plot(num_qubits, num_qubits**1.8, "o", label="Distance = 0.08")
plt.plot(num_qubits, num_qubits**2, "o", label="Distance = 0.04")
plt.plot(num_qubits, num_qubits**2.2, "o", label="Distance = 0.02")

plt.plot(num_qubits, num_qubits**1.8, ":", alpha=0.5)
plt.plot(num_qubits, num_qubits**2, ":", alpha=0.5)
plt.plot(num_qubits, num_qubits**2.2, ":", alpha=0.5)

plt.xlabel("Number of qubits $q$")
plt.ylabel("$|P|$")

plt.title("Required amount of Pauli strings to achieve desired distance from ER spectrum")
plt.legend()
plt.show()
```
